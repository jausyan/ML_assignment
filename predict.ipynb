```python
import pandas as pd
```

```python
df = pd.read_csv('salary_datasets.csv')
df.head()
```

```python
print(df.shape)
```

```python
# linear regression simple to predict salary patern
# 1. split data into x and y
x = df.drop('salary', axis=1)
y = df['salary']  
```

```python
# 2. split data into train and test
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# 3. train the model
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
# 4. predict the model
y_pred = model.predict(x_test)
print(y_pred)
```

```python
# 5. evaluate the model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print('Mean Absolute Error:', mae)
print('Mean Squared Error:', mse)
print('R2 Score:', r2)
```

```python
# prdict salary for 10 year experience
experience = 2
predicted_salary = model.predict([[experience]])
print('Predicted Salary for 10 years of experience:', predicted_salary[0])
```

```python
# predict from csv file
new_data = pd.read_csv('test_data.csv')
new_x = new_data.drop('salary', axis=1)
new_y_pred = model.predict(new_x)
print('Predicted Salaries for new data:', new_y_pred)
```

```python
# visualize all data
import matplotlib.pyplot as plt
plt.scatter(df['year of experience'], df['salary'], color='blue', label='Actual Data')
plt.plot(df['year of experience'], model.predict(df[['year of experience']]), color='red', label='Regression Line')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Experience vs Salary')
plt.legend()
plt.show()
```
